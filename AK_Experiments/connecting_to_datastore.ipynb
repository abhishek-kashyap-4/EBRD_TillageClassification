```python
import re
from azure.ai.ml import MLClient
from azure.identity import DefaultAzureCredential

def parse_azureml_uri(azureml_uri: str, ml_client: MLClient):
    """
    Parses an azureml:// datastore URI and returns the native az:// storage URI
    and account details.
    """
    # Regex to extract datastore name and relative path
    pattern = r"azureml://subscriptions/[^/]+/resourcegroups/[^/]+/workspaces/[^/]+/datastores/([^/]+)/paths/(.+)"
    match = re.match(pattern, azureml_uri)
    
    if not match:
        raise ValueError("Invalid azureml:// datastore URI format.")
        
    datastore_name, relative_path = match.groups()
    
    # Retrieve datastore properties from Azure ML
    datastore = ml_client.datastores.get(datastore_name)
    
    account_name = datastore.account_name
    container_name = datastore.container_name
    
    # Construct native az:// URI
    az_uri = f"az://{container_name}/{relative_path}"
    
    return {
        "az_uri": az_uri,
        "account_name": account_name,
        "container_name": container_name,
        "relative_path": relative_path
    }

# --- Usage Example ---
ml_client = MLClient.from_config(credential=DefaultAzureCredential())

azureml_uri = "azureml://subscriptions/f3ad55ae-d72a-4cce-aa07-ccefed436a79/resourcegroups/rg-rise-westeurope/workspaces/mlw-rise-westeurope/datastores/rise_data/paths/shared/ebrd_tillage/s2/stac_inventory_items.geojson"
info = parse_azureml_uri(azureml_uri, ml_client)

print("Native Storage URI:", info["az_uri"])
print("Storage Account:", info["account_name"])
print("Container:", info["container_name"])
```

```python
from adlfs import AzureBlobFileSystem
from azure.identity import DefaultAzureCredential

# Pass account name from Method 1
fs = AzureBlobFileSystem(
    account_name=info["account_name"],
    credential=DefaultAzureCredential()
)

# 1. Check if path exists
path_to_check = f"{info['container_name']}/{info['relative_path']}"
print("Exists:", fs.exists(path_to_check))

# 2. List contents inside your Zarr folder (e.g., .zgroup, .zattrs, array folders)
print("Directory contents:")
print(fs.ls(path_to_check))
```
