```python
from pathlib import Path
import geopandas as gpd
from datetime import datetime, timedelta

# Build an absolute path pointing to ~/data
# Path.home() resolves to your root user directory (e.g., /home/yourname)
data_dir = Path.home() / "EBRD_Tillage" /"data" / "input"

print(data_dir)
# Optional: If you want to use your script's logic to automatically 
# find the shapefile inside the ~/data folder:
def resolve_fields_path(path):
    p = Path(path)
    if not p.is_dir():
        return p
    for pattern in ("*.shp", "*.gpkg", "*.geojson"):
        matches = sorted(p.glob(pattern))
        if matches:
            return matches[0]
    raise FileNotFoundError(f"No vector file found in: {p}")

shapefile_path = resolve_fields_path(data_dir)
print(f"Reading shapefile from: {shapefile_path}")

# Load the shapefile with GeoPandas
gdf = gpd.read_file(shapefile_path)

# Quick check
print(f"Successfully loaded {len(gdf)} rows.")
print(gdf.head())
```

```python
sample_field_gdf = gdf.iloc[0:1]
sample_field = sample_field_gdf.iloc[0]
start = sample_field.Beginning 
start = datetime.strptime(start, '%Y/%m/%d %H:%M:%S.%f')

end = sample_field['End date']
end = datetime.strptime(end, '%Y/%m/%d %H:%M:%S.%f')

start = start - timedelta(days=5)
end = end + timedelta(days=5) 
print('Padded Dates',start,end)



sample_field_gdf.to_file("/home/azureuser/EBRD_Tillage/data/interim/sample.geojson", driver="GeoJSON")


```
